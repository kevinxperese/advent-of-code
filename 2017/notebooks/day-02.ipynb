## Parse the Input Data

```python
def parse(filename):
    """Parse puzzle input data."""
    with open(f'../inputs/{filename}.txt') as f:
        rows = []
        for line in f.readlines():
            rows.append([int(n) for n in line.strip().split('\t')])
    return rows
```

## Part 1
---

```python
def solve(spreadsheet):
    checksum = 0
    for row in spreadsheet:
        checksum += max(row) - min(row)
    return checksum
```

### Run on Test Data

```python
print(solve(parse("day-02_test")) == 18)
```

### Run on Input Data

```python
solve(parse("day-02"))
```

## Part 2
---

```python
from itertools import permutations
```

```python
def solve2(spreadsheet):
    checksum = 0
    for row in spreadsheet:
        for p in permutations(row, 2):
            if p[0] % p[1] == 0:
                checksum += p[0] / p[1]
                break
    return int(checksum)
```

### Run on Test Data

```python
solve2(parse("day-02_test2")) == 9
```

### Run on Input Data

```python
solve2(parse("day-02"))
```
